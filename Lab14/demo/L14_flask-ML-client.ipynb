# REST API by flask (Client)
This notebook is a simple example of how to use REST API by flask.

```python
# Import the necessary libraries
import requests, json, sys
```

## Create a post request
The following code is an example of how to create a post request to the server.

```python
def predict_diabetes(BMI, Age, Glucose, url):
    # Construct the data payload
    # And convert it to json
    data = {"BMI":BMI, "Age":Age, "Glucose":Glucose}
    data_json = json.dumps(data)

    # Set the content type in the headers
    headers = {'Content-type':'application/json'}

    # Make a POST request to the model server
    response = requests.post(url, data=data_json, headers=headers)

    # If the request was successful, return the result
    if response.status_code == 200: 
        result = json.loads(response.text)
        return result
    else:
        return None
```

# Try it yourself
You can try to run the following code in your local machine.

```python
url = "URL_TO_YOUR_MODEL_SERVER"  # Replace with your model server URL
BMI = 25
Age = 25
Glucose = 200

# Call the function and print the result
result = predict_diabetes(BMI, Age, Glucose, url)
print(result)
```
