# REST API by flask (Client)
This notebook is a simple example of how to use REST API by flask.

```python
# Import the necessary libraries
import requests, json, sys
```

## Create a post request
The following code is an example of how to create a post request to the server. The server will return the sum of the two numbers.

```python
def square_number_post(number, url):

    # Prepare the data payload
    data = {"number": number}
    
    # Set the headers for JSON content
    headers = {"Content-Type": "application/json"}
    
    # Send POST request
    response = requests.post(url, data=json.dumps(data), headers=headers)
    
    # Check if request was successful
    if response.status_code == 200:
        result = response.json()
        print(f"POST Method Result: {number} squared equals {result['result']}")
        return result
    else:
        print(f"Error: Server returned status code {response.status_code}")
        print(f"Response: {response.text}")
        return None
```

# Try it yourself
You can try to run the following code in your local machine.

```python
url = "http://localhost:5000/square"
number = 50
square_number_post(number, url)
```
