# REST API by flask (Server)
This notebook is a simple example of how to create a REST API using Flask. 

````python
# Import the Flask class from the flask module
from flask import Flask, request, jsonify
````

## Initialize the Flask app
Then, we need to initialize the Flask app and create a route that will call the function we created.

````python
# Create a new instance of the Flask class
app = Flask(__name__)

# Define a route for the default URL and its behavior
@app.route('/square', methods=['POST'])
def square_number():
    
    # Get the data from the request
    # The data is a dictionary with the key 'number'
    data = request.get_json()
    
    # Get the number from the dictionary and convert it to a float
    # Then calculate the square of the number
    number = float(data['number'])
    result = number ** 2
    
    # Create a dictionary with the key 'result' and the value of the square
    outdata = {'result': result}    

    # Return the dictionary as a JSON string
    return jsonify(outdata)
````

## Run the app
Finally, we need to run the app.

````python
app.run(host='0.0.0.0', port=5000, debug=False)
````
